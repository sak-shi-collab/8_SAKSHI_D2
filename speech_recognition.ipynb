```python
!pip install pyttsx3
!apt-get update
!apt-get install -y espeak-ng
import pyttsx3

engine = pyttsx3.init()

```

```python
text = """
Hello sakshi.
Welcome to the Artificial Intelligence laboratory.
This is a sample voice recording for the Speech to Text and Text to Speech experiment.
Have a nice day.
"""
```

```python
engine.setProperty('rate', 150)
engine.setProperty('volume', 5.0)

engine.save_to_file(text, 'hello_sakshi.wav')
engine.runAndWait()

print("WAV file created successfully.")
```
